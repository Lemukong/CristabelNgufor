## 1. Introduction

Shadow Protocol. Bamenda Edition plunges you into this hidden world, where every shadow could conceal a clue, and every connection could be a trap.

You are about to step into the shoes of Detective Cristabel Morgan, an investigator whose intuition and sharp mind are her greatest weapons, and Godson Ray, a hacker whose skills allow him to navigate networks that most fear to touch. ##4. Setup (dependencies and helpers)

````python
# Setup: imports and helper functions
import random
import time

# Optional color support
try:
    from colorama import Fore, Style, init
    init(autoreset=True)
    COLOR = True
except Exception:
    COLOR = False

def cprint(text, color=None, delay=0):
    """Print with optional color and small delay for effect."""
    if COLOR and color:
        col = getattr(Fore, color.upper(), "")
        try:
            print(col + text + Style.RESET_ALL)
        except Exception:
            print(text)
    else:
        print(text)
    if delay:
        time.sleep(delay)
````

###5. ASCII Art & Scenes

````python
# ASCII Art and Scenes (helpers)
def display_banner():
    art = r"""
  ____  _               _                ____            _       _
 / ___|| |__   ___  ___| | _____ _ __   |  _ \ ___  __ _(_)_ __ | |_ ___
 \___ \| '_ \ / _ \/ __| |/ / _ \ '__|  | |_) / _ \/ _` | | '_ \| __/ __|
  ___) | | | |  __/ (__|   <  __/ |     |  _ <  __/ (_| | | | | | |_\__ \
 |____/|_| |_|\___|\___|_|\_\___|_|     |_| \_\___|\__, |_|_| |_|\__|___/
                                                   |___/
    """
    cprint(art, "cyan")

def ascii_crime_scene():
    art = r"""
+--------------------------------------------+
|   CRIME SCENE: Shattered Research Tablet   |
|   Location: Up Station residence (Bamenda) |
|                                            |
|     [ shattered datapad near the body ]    |
+--------------------------------------------+
    """
    cprint(art, "magenta")

def ascii_bar_scene():
    art = r"""
+--------------------------------------------+
|               THE MARKET BAR               |
|   A busy spot near Commercial Avenue       |
|   A shadow hands over a small data chip    |
+--------------------------------------------+
    """
    cprint(art, "magenta")

def ascii_firewall():
    art = r"""
+--------------------------------------------+
|  FIREWALL: |||===[========]===|||          |
|  TRACE METER: [=====     ]                 |
+--------------------------------------------+
    """
    cprint(art, "yellow")

def ascii_true_ending():
    art = r"""
+============================================+
|   TRUE ENDING: SHADOW PROTOCOL EXPOSED     |
|  Cristabel and Godson broadcast the truth    |
|  Bamenda breathes a sigh of relief         |
+============================================+
    """
    cprint(art, "green")
````

###6. Detective Campaign

````python
# Detective Campaign: Cristabel (Bamenda)
def detective_start():
    cprint("\n🕵️ DETECTIVE CRISTABEL MORGAN — CASE FILE: BAMENDA-77", "yellow")
    cprint("A prominent researcher linked to Dr. Ngufor has been found dead at Up Station.\n", "magenta")
    player = {"clues": [], "trust": 0}
    crime_scene(player)

def crime_scene(player):
    ascii_crime_scene()
    cprint("Options: [inspect], [interview], [go to lab]")
    choice = input("> ").lower().strip()
    if choice == "inspect":
        cprint("Cristabel finds encrypted logs labeled: 'Project Umbra — Dr. Ngufor'.", "magenta")
        if "ngufor logs" not in player["clues"]:
            player["clues"].append("ngufor logs")
        crime_scene(player)
    elif choice == "interview":
        cprint("A street vendor says: 'I saw Godson arguing with him about stolen data near Commercial Avenue.'", "magenta")
        if "Godson argument" not in player["clues"]:
            player["clues"].append("Godson argument")
            player["trust"] += 1
        crime_scene(player)
    elif choice == "go to lab":
        lab_scene(player)
    else:
        cprint("Invalid choice. Try again.", "red")
        crime_scene(player)

def lab_scene(player):
    cprint("\n📍 DR. NGUFOR’S LAB — Near Ntamulung area, Bamenda.", "magenta")
    cprint("Options: [question ngufor], [search], [go to market bar]")
    choice = input("> ").lower().strip()
    if choice == "question ngufor":
        cprint("Dr. Ngufor: 'My neural core research could revolutionize healthcare... I would never harm anyone.'", "magenta")
        if "ngufor denial" not in player["clues"]:
            player["clues"].append("ngufor denial")
        lab_scene(player)
    elif choice == "search":
        cprint("Cristabel finds a hidden device labeled 'Neural Core — Prototype'.", "magenta")
        if "neural core" not in player["clues"]:
            player["clues"].append("neural core")
        lab_scene(player)
    elif choice == "go to market bar":
        bar_scene(player)
    else:
        cprint("Invalid choice. Try again.", "red")
        lab_scene(player)

def bar_scene(player):
    ascii_bar_scene()
    cprint("Options: [talk to Godson], [observe], [finish]")
    choice = input("> ").lower().strip()
    if choice == "talk to Godson":
        cprint("Godson: 'I didn't kill anyone. But I did access files that point to a cover-up.'", "magenta")
        if "Godson innocence" not in player["clues"]:
            player["clues"].append("Godson innocence")
        bar_scene(player)
    elif choice == "observe":
        cprint("Cristabel notices a stranger passing Godson a small, encrypted data chip.", "magenta")
        if "mystery chip" not in player["clues"]:
            player["clues"].append("mystery chip")
        bar_scene(player)
    elif choice == "finish":
        detective_ending(player)
    else:
        cprint("Invalid choice. Try again.", "red")
        bar_scene(player)

def detective_ending(player):
    cprint("\n🕵️ FINAL DECISION — Who does Cristabel accuse?", "yellow")
    cprint("1 — Godson (the hacker)\n2 — Dr. Ngufor\n3 — The unknown third party (corporate hit)", "yellow")
    choice = input("> ").strip()
    if choice == "3" and "mystery chip" in player["clues"]:
        cprint("\n✅ Cristabel uncovers evidence that a hired assassin was protecting corporate secrets.", "green")
        cprint("Her investigation points toward a larger conspiracy tied to Dr. Ngufor's research.", "green")
        cprint("\nSwitching to Hacker route to finish the story...", "green")
        hacker_start()
    elif choice == "1":
        cprint("\n❌ Wrong accusation — Godson was innocent. The trail goes cold.", "red")
        start_menu()
    elif choice == "2":
        cprint("\n❌ Dr. Ngufor had secrets but wasn't the killer. The conspirators slip away.", "red")
        start_menu()
    else:
        cprint("\nCase unresolved. Returning to main menu.", "yellow")
        start_menu()
````

###7. Hacker Campaign

````python
# Hacker Campaign: Godson (Bamenda)
def hacker_start():
    cprint("\n💻 GODSON RAY — TERMINAL ONLINE (Bamenda)", "yellow")
    cprint("Goal: Infiltrate the megacorp servers and retrieve data connected to Project Umbra.\n", "magenta")
    player = {"trace": 0, "data": 0, "tools": ["BruteForcer", "StealthMask"]}
    firewall(player)

def firewall(player):
    ascii_firewall()
    cprint(f"\n🔥 FIREWALL — Trace: {player['trace']}% — Data fragments: {player['data']}/3", "yellow")
    cprint("Options: [bruteforce], [bypass], [scan]", "yellow")
    choice = input("> ").lower().strip()
    if choice == "bruteforce":
        if random.random() < 0.6:
            cprint("Firewall cracked.", "magenta")
            player["data"] += 1
            data_center(player)
        else:
            cprint("Security detected your attempt!", "red")
            player["trace"] += 20
            check_trace(player)
    elif choice == "bypass":
        cprint("StealthMask activated. You bypass silently.", "magenta")
        player["data"] += 1
        data_center(player)
    elif choice == "scan":
        cprint("Scan found tool: ZeroDay exploit.", "magenta")
        if "ZeroDay" not in player["tools"]:
            player["tools"].append("ZeroDay")
        firewall(player)
    else:
        cprint("Invalid choice. Try again.", "red")
        firewall(player)

def data_center(player):
    cprint(f"\n📂 DATA CENTER — Trace: {player['trace']}% — Data fragments: {player['data']}/3", "yellow")
    cprint("Options: [decrypt], [download], [exit]", "yellow")
    choice = input("> ").lower().strip()
    if choice == "decrypt":
        if "ZeroDay" in player["tools"]:
            cprint("Decryption successful — you read files mentioning 'Neural Core' and 'Project Umbra'.", "magenta")
            player["data"] += 1
            player["trace"] += 5
        else:
            cprint("Decryption failed — trace increased.", "red")
            player["trace"] += 15
        check_trace(player)
    elif choice == "download":
        cprint("Download started — but it triggered an alert!", "red")
        player["data"] += 1
        player["trace"] += 20
        check_trace(player)
    elif choice == "exit":
        control_room(player)
    else:
        cprint("Invalid choice. Try again.", "red")
        data_center(player)

def control_room(player):
    cprint(f"\n⚠️ CONTROL ROOM — Trace: {player['trace']}% — Data fragments: {player['data']}/3", "yellow")
    cprint("Options: [upload], [erase], [disconnect]", "yellow")
    choice = input("> ").lower().strip()
    if choice == "upload":
        if player["data"] >= 3:
            ascii_true_ending()
            cprint("Cristabel receives the leaked files and the conspiracy unravels.", "green")
            cprint("\n🎉 TRUE ENDING — Justice for Bamenda.", "green")
        else:
            cprint("Not enough data fragments to form a complete leak.", "red")
            player["trace"] += 15
            check_trace(player)
    elif choice == "erase":
        cprint("Erasing traces... Trace reduced by 25%.", "magenta")
        player["trace"] = max(0, player["trace"] - 25)
        control_room(player)
    elif choice == "disconnect":
        cprint("Disconnecting safely. Returning to menu.", "yellow")
    else:
        cprint("Invalid choice. Try again.", "red")
        control_room(player)
    start_menu()

def check_trace(player):
    if player["trace"] >= 100:
        cprint("\n🚨 TRACE COMPLETE — Godson has been traced and captured.", "red")
        start_menu()
    elif player["data"] >= 3:
        control_room(player)
    else:
        firewall(player)
````

###8. Shared Ending & Runner

````python
# Shared Ending & Runner
def start_menu():
    display_banner()
    cprint("\nSHADOW PROTOCOL — BAMENDA EDITION", "cyan")
    print("1 — Play Detective Cristabel")
    print("2 — Play Hacker Godson")
    print("3 — Read instructions / credits")
    print("4 — Exit")
    choice = input("> ").strip()
    if choice == "1":
        detective_start()
    elif choice == "2":
        hacker_start()
    elif choice == "3":
        # show_instructions defined earlier
        print("\nINSTRUCTIONS: Type the commands shown in the brackets and press Enter.")
        input("\nPress Enter to return to menu...")
        start_menu()
    else:
        cprint("Goodbye.", "cyan")

# If you paste this notebook into Colab, run this cell to start the game.
if __name__ == "__main__":
    start_menu()
````
